```python
import numpy as np
from scipy.sparse import csr_matrix
import time
from IPython.display import clear_output

N = 400
step = 1
n_time = 2000

time_np = []
time_sci = []
size = []

for n in range(4, N, step) :
    matrix_np = np.zeros((n,n))
    for column in range(n) :
        indexs = np.random.randint(0,n-1,4)
        for index in indexs :
            matrix_np[index][column] = 1
    matrix_sci = csr_matrix(matrix_np)
    
    time_0 = time.time()

    for i in range(n_time) :
        np.dot(matrix_np, matrix_np)

    time_1 = time.time()

    for i in range(n_time) :
        matrix_sci.dot(matrix_sci)

    time_2 = time.time()

    time_np.append(time_1 - time_0)
    time_sci.append(time_2 - time_1)
    size.append(n)

    clear_output(wait=False)
    print(f"{n}\n")
```

```python
import matplotlib.pyplot as plt

plt.plot(size, time_np, "-b", label="dense matrix")
plt.plot(size, time_sci, "-r", label="sparse matrix")
plt.legend(loc="upper left")
plt.xlabel("Size of the matrix")
plt.ylabel("Computation time (s)")
plt.show
```
